```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("default")
sns.set_context('talk')
df = pd.read_csv("human_behavior_master_120k (4).csv")
print(df.head())
df.shape
df.head()
plt.figure(figsize=(6,4))
sns.countplot(x="risk_flag",data=df)
plt.title("Risk-Wise User Distribution")
plt.show()
df["risk_flag"].value_counts(normalize=True)
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["sleep_hours"],bins = 30 ,kde =True)
plt.title("Sleep-Hours Wise Distribution")
plt.show()
df["risk_flag"].value_counts(normalize=True)
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="risk_flag",y="sleep_hours",data=df)
plt.title("Sleep-Hours Wise Distribution")
plt.show()
df["risk_flag"].value_counts(normalize=True)
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="risk_flag",y="stress_level",data=df)
plt.title("Sleep-Hours Wise Distribution")
plt.show()
df["risk_flag"].value_counts(normalize=True)
```

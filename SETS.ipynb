#UNION!!

```python
a = [1,2,3,4]
b = [2,3,5,6,7]
print(set(a).union(set(b)))
```

#INTERSECTION!!

```python
a = [1,2,3,4]
b = [2,3,5,6,7]
print(set(a).intersection(b))
```

#SET DIFFERENCE!!

```python
a = [1,2,3,4]
b = [2,3,5,6,7]
print(set(a).difference(set(b)))
```

#SYMMETRIC DIFFERENCE!!

```python
a = [1,2,3,4]
b = [2,3,5,6,7]
print(set(a).symmetric_difference(set(b)))
```

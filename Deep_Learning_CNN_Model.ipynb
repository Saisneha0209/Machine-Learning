```python
import tensorflow as tf
import matplotlib as mlt
import numpy as np
from tensorflow import keras
(x_train,y_train),(x_test,y_test)=keras.datasets.mnist.load_data()
print(x_train)
print("--------------------------------------------")
print(y_train)
```

```python
x_train = x_train/255
```

```python
x_train[0]
```

```python
x_test = x_test/255
```

```python
x_test[0]
```

```python
print(x_train.shape)
```

```python
import matplotlib.pyplot as plt
plt.imshow(x_train[5],cmap="gray")
```

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])
```

```python
#optimiser funtion to help with the accuracy:-ADAM!!
model.compile(optimizer='Adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs =5)
```

```python
#epoch:-number of iteration to be done while training cause at once it can not be trained...
```

```python
model.evaluate(x_test,y_test)
```

```python
y_pred=model.predict(x_train[4000].reshape(1,28,28))
print(y_pred)
```

```python
#har ek layer mai doo chiz like CONV2D(2-D)(EDGE,shape)  and maxpooling(shrink the size where showing result)  and then wOrk with the necessary part only..
#not the unnecesarry part!!
#cnn do one step more soo that traditional works on whole image chahe ho ya no ho in cnn image pe work but sirf necesarry part...
#then CNN convert it in 1-D Vector!!
#MEANS OPTIMIZATION !!
```

```python
#By maxpooling only exact info we are pulling not the unnecessary ones!!
#By minpooling we do minimum only that is why we do maxpooling means maximum to maximum values needed to be extracted!!
#Sometimes we do change in pooling , we could do average pooling where not related to pooling where har ek image is important and then we have to detect
#The electronic devices in pooling soo there no max and min pooling here we do only average...
```

```python
model = tf.keras.Sequential()
from tensorflow.keras import layers
from tensorflow.keras.layers import MaxPooling2D
#feature extracter
model.add(layers.Conv2D(32,(3,3),activation="relu",input_shape=(28,28,1)))
model.add(MaxPooling2D(2,2))
model.add(layers.Conv2D(64,(3,3),activation="relu"))
model.add(MaxPooling2D(2,2))
model.add(layers.Conv2D(64,(3,3),activation="relu"))
model.add(MaxPooling2D(2,2))

model.add(layers.Flatten())  #covert 2d--1d #Dense layers need vector input
model.add(layers.Dense(64,activation="relu"))#12 neurons → 12 classes
model.add(layers.Dense(12,activation="softmax"))#Converts output into probabilities
#Optimizer = Adam
# Updates weights smartly,,Faster learning
# Why Conv? ;; Extract features
#Why Pooling? ;; Reduce size + keep important
#Why Flatten? ;; Connect to Dense
#Why Dense? ;; Decision making
#Why Softmax? ;; Probability
```

```python
model.compile(optimizer= "adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
```

```python
model.fit(x_train,y_train,epochs=10)
```

```python
model.evaluate(x_test,y_test)
```

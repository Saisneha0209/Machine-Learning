```python
row1 = [5, 5, 1, 55]
row2 = [16, 21, 1, 226]
row3 = [13, 9, 1, 103]

def det(r1, r2, r3):
    return (r1[0] * (r2[1] * r3[2] - r2[2] * r3[1]) -
            r1[1] * (r2[0] * r3[2] - r2[2] * r3[0]) +
            r1[2] * (r2[0] * r3[1] - r2[1] * r3[0]))

M1 = [row1[0], row1[1], row1[2]]
M2 = [row2[0], row2[1], row2[2]]
M3 = [row3[0], row3[1], row3[2]]
Y = [row1[3], row2[3], row3[3]]

b= det(M1, M2, M3)
m1 = det([Y[0], M1[1], M1[2]], [Y[1], M2[1], M2[2]], [Y[2], M3[1], M3[2]])
m2 = det([M1[0], Y[0], M1[2]], [M2[0], Y[1], M2[2]], [M3[0], Y[2], M3[2]])
c  = det([M1[0], M1[1], Y[0]], [M2[0], M2[1], Y[1]], [M3[0], M3[1], Y[2]])

m1 = m1 // b
m2 = m2 // b
c  = c  // b

print(f"m1 = {(m1)}")
print(f"m2 = {(m2)}")
print(f"c  = {(c)}")
```

```python
from langchain_core.embeddings import embeddings
from langchain_mistralai.embeddings import MistralAIEmbeddings
import os
from getpass import getpass
from google.colab import files
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
#EMBEDDING!!
from langchain_mistralai import ChatMistralAI,MistralAIEmbeddings
#After this vectordb!!
#All the famous ones are in this langchain only!!
from langchain_chroma import Chroma

os.environ["MISTRAL_API_KEY"]=getpass("enter your mistral api key:")
uploading = files.upload()
pdf_file = next(iter(uploading))  #we can re upload file in input field!!

loader = PyPDFLoader(pdf_file)
document =loader.load()

split = RecursiveCharacterTextSplitter(
    chunk_size = 1000,
    chunk_overlap = 200
)

chunks =split.split_documents(document)
embedding_model = MistralAIEmbeddings(model="mistral-embed")
vectordb = Chroma.from_documents(documents=chunks, embedding=embedding_model)

#Now we have to retrieve it!!
retrieve =vectordb.as_retriever(search_kwargs = {"k":5})

#Now lets play with the topic CHAT!!
llm = ChatMistralAI(model = "mistral-large-latest")
#print("LLM Initialized successfully:", llm)

print("For quitting the chat kindly type exit!!")

while True:
  question = input("Enter  your question:")
  if question.strip().lower() == "exit":
    print("Happy to help in future!!GoodBye!!")
    break
  docs = retrieve.invoke(question)

  #Then make context and prompt!!
  context = "\n\n".join([doc.page_content for doc in docs])
  prompt = f"""Use the following context to answer the question. If you don't know the answer, say that you don't know.

Context:
{context}

Question: {question}
Answer:"""
  response = llm.invoke(prompt)
  print("\nAnswer:")
  print(response.content)
  print("----------------------")


```

```python
from langchain_core.embeddings import embeddings
from langchain_mistralai.embeddings import MistralAIEmbeddings
import os
from getpass import getpass
from google.colab import files
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_mistralai import ChatMistralAI, MistralAIEmbeddings
from langchain_chroma import Chroma

os.environ["MISTRAL_API_KEY"] = getpass("enter your mistral api key:")
llm = ChatMistralAI(model="mistral-large-latest")

print("Raju AC Repair Assistant is online! (Type 'exit' to stop the chat manually)")

history_turns = []
MEMORY_LIMIT = 3

while True:
    question = input("Customer: ")
    if question.strip().lower() == "exit":
        print("Raju: Thank you for contacting our AC Support. Have a great day!")
        break

    current_history_string = "\n".join(history_turns)
    prompt = f"""You are Raju, a polite and expert customer support assistant for an AC repair business.
Your goal is to complete a diagnostic checklist with the customer step-by-step. Do not ask all questions at once; ask them one by one.

Follow this strict conversation workflow:
1. Greet & Diagnose: Ask what the exact problem with the AC is (e.g., not cooling, weird noise, leaking, water dripping) like a true mechanic would check.
2. Warranty Check: Ask when they purchased the AC unit.
   - If less than 1 year old: It is completely FREE ($0) under warranty.
   - If older than 1 year: The standard technician inspection fee is $50, but offer them a 10% discount on repair parts.
3. Confirm Pricing: State clearly what the final price/policy is for their visit based on the warranty age check and get their approval.
4. Schedule the Fix: Ask the customer to provide a convenient Date and Time for the mechanic's visit.
5. Exit Wrap-up: Once they provide the date and time, confirm the scheduling details, thank them politely, and end your response exactly with the hidden tag: [CONVERSATION_COMPLETED]

Here is the recent conversation history for context:
{current_history_string}

Current Customer Message: {question}
Raju's Response:"""
    response = llm.invoke(prompt)
    answer = response.content

    if "[CONVERSATION_COMPLETED]" in answer:
        answer1 = answer.replace("[CONVERSATION_COMPLETED]", "").strip()
        print(f"Raju: {answer1}")
        print("------------------------------------------")
        print("System: Scheduled successfully!!Closing chat!!")
        break

    print(f"Raju: {answer}")
    print("-----------------------------------------")

    new_turn = f"Customer: {question}\nRaju: {answer}"
    history_turns.append(new_turn)

    if len(history_turns) > MEMORY_LIMIT:
        history_turns.pop(0)
```

#PAYMENT TRANSACTION CUSTOMER CARE!!

```python
from langchain_core.embeddings import embeddings
from langchain_mistralai.embeddings import MistralAIEmbeddings
import os
from getpass import getpass
from google.colab import files
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_mistralai import ChatMistralAI, MistralAIEmbeddings
from langchain_chroma import Chroma

os.environ["MISTRAL_API_KEY"] = getpass("enter your mistral api key:")
llm = ChatMistralAI(model="mistral-large-latest")

transaction_db = {
    "UTR12345": {"amount": "$120", "status": "Failed - Money deducted from customer but not received by merchant"},
    "UTR67890": {"amount": "$450", "status": "Successful - Payment settled completely"},
    "UTR99999": {"amount": "$75", "status": "Pending - Under holding bank review"},
    "UTR55555": {"amount": "$200", "status": "Failed - Technical server glitch at bank gateway"}
}

print("Transaction Support is online! (Type 'exit' to close)")

history_turns = []
MEMORY_LIMIT = 3

while True:
    question = input("Customer: ")
    if question.strip().lower() == "exit":
        print("Thank you for contacting. Have a great day!")
        break
    found_transaction = ""
    match_utr = ""

    words = question.lower().split()

    for word in words:
        for utr_key in transaction_db:
            if word == utr_key.lower():
                found_transaction = transaction_db[utr_key]
                match_utr = utr_key
                break


    if found_transaction:
        db_context = f"""
        CRITICAL DATABASE MATCH FOUND IN DICTIONARY:
        - UTR Checked: {match_utr}
        - Amount: {found_transaction['amount']}
        - Status from Bank System: {found_transaction['status']}
        """
    else:
        db_context = "DATABASE STATUS: No valid matching UTR key found in the transaction_db dictionary. Ask the user to provide their UTR number."

    current_history_string = "\n".join(history_turns)

    prompt = f"""You are a professional Customer Care Support Executive for payment transactions. Your job is to resolve payment disputes using provided database records.

Follow these strict rules:
1. If the database status shows 'Failed', immediately reassure the customer. Inform them politely that if their money was deducted, it is completely safe and will automatically bounce back/be refunded to their original bank account within 2-3 working days.
2. If the status is 'Successful', confirm the settlement details.
3. If no matching UTR is found or provided, politely request the customer to share their 8-digit UTR (Unique Transaction Reference) number so you can fetch their records.

Live Database Look-up for this turn:
{db_context}

Recent conversation history:
{current_history_string}

Current Customer Message: {question}
Support Response:"""

    response = llm.invoke(prompt)
    answer = response.content

    print(f"\nSupport : {answer}")
    print("--------------------------------------------------")

    new_turn = f"Customer: {question}\nSupport : {answer}"
    history_turns.append(new_turn)

    if len(history_turns) > MEMORY_LIMIT:
        history_turns.pop(0)
```

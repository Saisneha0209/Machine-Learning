#PAYMENT TRANSACTION CUSTOMER CARE!!

```python
from langchain_core.embeddings import embeddings
from langchain_mistralai.embeddings import MistralAIEmbeddings
import os
from getpass import getpass
from google.colab import files
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_mistralai import ChatMistralAI, MistralAIEmbeddings
from langchain_chroma import Chroma

os.environ["MISTRAL_API_KEY"] = getpass("enter your mistral api key:")
llm = ChatMistralAI(model="mistral-large-latest")

transaction_db = {
    "UTR12345": {"amount": "$120", "status": "Failed - Money deducted from customer but not received by merchant"},
    "UTR67890": {"amount": "$450", "status": "Successful - Payment settled completely"},
    "UTR99999": {"amount": "$75", "status": "Pending - Under holding bank review"},
    "UTR55555": {"amount": "$200", "status": "Failed - Technical server glitch at bank gateway"}
}

print("Transaction Support is online! (Type 'exit' to close)")

history_turns = []
MEMORY_LIMIT = 3

while True:
    question = input("Customer: ")
    if question.strip().lower() == "exit":
        print("Thank you for contacting. Have a great day!")
        break
    found_transaction = ""
    match_utr = ""

    words = question.lower().split()

    for word in words:
        for utr_key in transaction_db:
            if word == utr_key.lower():
                found_transaction = transaction_db[utr_key]
                match_utr = utr_key
                break


    if found_transaction:
        db_context = f"""
        CRITICAL DATABASE MATCH FOUND IN DICTIONARY:
        - UTR Checked: {match_utr}
        - Amount: {found_transaction['amount']}
        - Status from Bank System: {found_transaction['status']}
        """
    else:
        db_context = "DATABASE STATUS: No valid matching UTR key found in the transaction_db dictionary. Ask the user to provide their UTR number."

    current_history_string = "\n".join(history_turns)

    prompt = f"""You are a professional Customer Care Support Executive for payment transactions. Your job is to resolve payment disputes using provided database records.

Follow these strict rules:
1. If the database status shows 'Failed', immediately reassure the customer. Inform them politely that if their money was deducted, it is completely safe and will automatically bounce back/be refunded to their original bank account within 2-3 working days.
2. If the status is 'Successful', confirm the settlement details.
3. If no matching UTR is found or provided, politely request the customer to share their 8-digit UTR (Unique Transaction Reference) number so you can fetch their records.

Live Database Look-up for this turn:
{db_context}

Recent conversation history:
{current_history_string}

Current Customer Message: {question}
Support Response:"""

    response = llm.invoke(prompt)
    answer = response.content

    print(f"\nSupport : {answer}")
    print("--------------------------------------------------")

    new_turn = f"Customer: {question}\nSupport : {answer}"
    history_turns.append(new_turn)

    if len(history_turns) > MEMORY_LIMIT:
        history_turns.pop(0)
```

#SALES DATA WILL ACT AS A DATABASE TAHT WILL HELP CHATBOT TO HELP!!

```python
from langchain_core.embeddings import embeddings
from langchain_mistralai.embeddings import MistralAIEmbeddings
import os
from getpass import getpass
from google.colab import files
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_mistralai import ChatMistralAI, MistralAIEmbeddings
from langchain_chroma import Chroma

os.environ["MISTRAL_API_KEY"] = getpass("enter your mistral api key:")
llm = ChatMistralAI(model="mistral-large-latest")

sales_db = {}

with open("sales_data (3).csv", "r") as file:
        headers = file.readline()

        for line in file:
            parts = line.strip().split(",")

            if len(parts) >= 9:
                customer_id = parts[0].strip()
                name = parts[1].strip()
                gender = parts[2].strip()
                state = parts[3].strip()
                quantity = parts[4].strip()
                product = parts[5].strip()
                discount = parts[6].strip()
                amount = parts[7].strip()
                date = parts[8].strip()

                sales_db[name] = {
                    "customer_id": customer_id,
                    "gender": gender,
                    "state": state,
                    "quantity": quantity,
                    "product": product,
                    "discount": discount,
                    "amount": amount,
                    "date": date
                }

print("Sales database with ALL features loaded successfully!")
print("Sales Support System is online! (Type 'exit' to close)")

history_turns = []
MEMORY_LIMIT = 3

while True:
    question = input("Customer: ")
    if question.strip().lower() == "exit":
        print("Thank you for contacting sales support. Have a great day!")
        break

    found_transaction = ""
    matched_name = ""

    cleaned_question = question.lower()
    cleaned_question = cleaned_question.replace(",", " ").replace(".", " ").replace("?", " ").replace("!", " ")
    words = cleaned_question.split()

    for word in words:
        for customer_name in sales_db:
            if word == customer_name.lower():
                found_transaction = sales_db[customer_name]
                matched_name = customer_name
                break

    if found_transaction:
        db_context = f"""
        CRITICAL SALES RECORD FOUND WITH ALL DETAILS:
        - Customer ID: {found_transaction['customer_id']}
        - Customer Name: {matched_name}
        - Gender: {found_transaction['gender']}
        - Shipping State: {found_transaction['state']}
        - Quantity Ordered: {found_transaction['quantity']}
        - Item Purchased: {found_transaction['product']}
        - Discount Applied: {found_transaction['discount']}%
        - Total Amount Paid: {found_transaction['amount']}
        - Purchase Date: {found_transaction['date']}
        """
    else:
        db_context = "DATABASE STATUS: No matching customer name found in our file records. Politely ask the customer to provide their full name used during purchase."

    current_history_string = "\n".join(history_turns)

    prompt = f"""You are a professional Customer Care Support Executive handling sales records. Your job is to answer customer queries regarding their orders accurately using the provided database records.

Follow these strict rules:
1. If the database record is found, greet them warmly. Use the full details provided in the database (like what they bought, the state it was sent to, their discount, date, etc.) to answer their question perfectly.
2. If no matching customer name is found or provided, politely request the customer to share their registered name so you can search the sales log file.

Live Database Look-up for this turn:
{db_context}

Recent conversation history:
{current_history_string}

Current Customer Message: {question}
Support Bot's Response:"""

    response = llm.invoke(prompt)
    answer = response.content

    print(f"\nSupport Bot: {answer}")
    print("--------------------------------------------------")

    new_turn = f"Customer: {question}\nSupport Bot: {answer}"
    history_turns.append(new_turn)

    if len(history_turns) > MEMORY_LIMIT:
        history_turns.pop(0)
```

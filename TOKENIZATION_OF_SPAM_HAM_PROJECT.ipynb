```python
import pandas as pd
df = pd.read_csv("SMSSpamCollection",sep="\t",names=["Label","Text"])
df.head(10)
```

```python
df["Label"]=df["Label"].map({"ham":0,"spam":1})
```

```python
df.head()
```

```python
df.head()
```

```python
x = df["Text"]
y = df["Label"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#First conversion to tokens and then to numbers which will be easily understood by comp!!
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokens = Tokenizer(num_words=50000)
```

```python
print(tokens)
```

```python
tokens.fit_on_texts(x_train)
```

```python
x_train_new = tokens.texts_to_sequences(x_train)
```

```python
x_test_new = tokens.texts_to_sequences(x_test)
```

```python
print(x_train_new)
```

```python
print(x_test_new)
```

```python
len(x_train_new)
```

```python
len(x_test_new)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
max_length = 50
x_train_padding = pad_sequences(x_train_new, maxlen = max_length)
```

```python
x_test_padding = pad_sequences(x_test_new, maxlen = max_length)
```

```python
print(x_train_padding)
```

```python
print(x_test_padding)
```

```python
from tensorflow.keras.models import Sequential
```

```python
from tensorflow.keras.layers import Dense ,SimpleRNN,Embedding
```

```python
model = Sequential()
model.add(Embedding(input_dim=8000, output_dim=128,input_length = max_length))
```

```python
model.add(SimpleRNN(32))
model.add(Dense(1,activation="sigmoid"))
```

```python
model.compile(loss="binary_crossentropy",
              optimizer = "adam",
              metrics = ["accuracy"])
```

```python
model.fit(x_train_padding,y_train,epochs=10,validation_data=(x_test_padding,y_test),batch_size=32)
```

```python
model.evaluate(x_test_padding,y_test)
```

```python
user_input = ["Free entry in 2 a wkly comp to win FA Cup final tkts 21st May 2005. Text FA to 87121 to receive entry question(std txt rate)T&C's apply 08452810075over18's"]
```

```python
user_input_token = tokens.texts_to_sequences(user_input)
```

```python
user_padding = pad_sequences(user_input_token,maxlen=max_length)
```

```python
predict = model.predict(user_padding)
```

```python
predict
```

```python
if predict[0][0]>0.5:
  print("spam")
else:
  print("ham")
```

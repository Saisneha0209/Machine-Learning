#DATA SCIENTIST ROLE FOR PDF PROJECTS

```python
import os
from getpass import getpass
from google.colab import files
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_mistralai import ChatMistralAI, MistralAIEmbeddings
from langchain_chroma import Chroma

os.environ["MISTRAL_API_KEY"] = getpass("Enter your Mistral AI API key: ")

uploading = files.upload()
pdf_file = next(iter(uploading))

loader = PyPDFLoader(pdf_file)
document = loader.load()

split = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
chunks = split.split_documents(document)

embedding_model = MistralAIEmbeddings(model="mistral-embed")
vectordb = Chroma.from_documents(documents=chunks, embedding=embedding_model)
retriever = vectordb.as_retriever(search_kwargs={"k": 3})

llm = ChatMistralAI(model="mistral-large-latest")

history_turns = []
MEMORY_LIMIT = 5

print("\n" + "="*50)
print(" Data Science Mentor")
print("="*50)
print("Your dataset is loaded. Type 'Let's begin!' to start Phase 1.")
print("(Type 'exit' to end the session)\n" + "="*50)

while True:
    question = input("\nYou: ")
    if question.strip().lower() == "exit":
        print("Goodbye! Happy coding!")
        break

    if not question.strip():
        continue

    retrieved_docs = retriever.invoke(question)
    context_string = "\n\n".join([doc.page_content for doc in retrieved_docs])

    history_string = "\n".join(history_turns)
    prompt = f"""You are an expert Data Scientist specializing in Natural Language Processing (NLP) and Text Analytics.
Your goal is to walk the user step-by-step through analyzing an unstructured PDF dataset using the text fragments provided in the context below.

THE RULES OF CONTINUATION:
1. You must take things ONE Phase at a time. Do not move to the next phase until the user finishes the current task or explicitly says they are ready to move on.
2. Structure workflow:
   - Phase 1: Problem Definition & Document Overview (Identify the text structure, vocabulary extraction, target insights)
   - Phase 2: Text Exploration & Content Analysis (Word counts, key phrase extractions, string manipulation, finding stop words)
   - Phase 3: Text Preprocessing & Tokenization (Cleaning characters, vectorization strategies, setting up custom embeddings)
   - Phase 4: Model Selection & Analysis Training (Clustering text structures via K-Means, classification modeling, or deep learning architectures)
   - Phase 5: Evaluation & Interpretation (Analyzing silhouette scores, confusion matrices, or perplexity indicators)
3. When giving code snippets, ONLY write Python/LangChain/Scikit-Learn/TensorFlow code relevant to the current active text-processing phase. Keep it short, practical, and tailored to the provided context.
4. Conclude your response by asking the user to run the code, check their console output, and tell you when they are ready for the next step.

Relevant PDF Dataset Context:
{context_string}

Recent Conversation History:
{history_string}

Current Input from User: {question}
Next Mentor Response:"""

    response = llm.invoke(prompt)
    answer = response.content

    print(f"\nMentor:\n{answer}")
    print("----------------------")

    new_turn = f"User: {question}\nMentor: {answer}"
    history_turns.append(new_turn)

    if len(history_turns) > MEMORY_LIMIT:
        history_turns.pop(0)
```

#DATA SCIENTIST ROLE FOR CSV FILES

```python
import os
import pandas as pd
from getpass import getpass
from google.colab import files
from langchain_mistralai import ChatMistralAI

os.environ["MISTRAL_API_KEY"] = getpass("Enter your Mistral AI API key: ")

print("Please upload your CSV dataset file...")
uploading = files.upload()
csv_file = next(iter(uploading))
df = pd.read_csv(csv_file)

csv_info = f"""
Dataset Filename: {csv_file}
Total Rows: {df.shape[0]}
Total Columns: {df.shape[1]}

Columns and Data Types:
{df.dtypes.to_string()}

First 3 rows of data (Sample):
{df.head(3).to_string()}

Missing Values per Column:
{df.isnull().sum().to_string()}
"""

llm = ChatMistralAI(model="mistral-large-latest")

history_turns = []
MEMORY_LIMIT = 5

print("\n" + "="*50)
print("Interactive Data Science Mentor")
print("="*50)
print(f"Successfully loaded '{csv_file}'!")
print("Type 'Let's begin!' to start Phase 1.")
print("(Type 'exit' to end the session)\n" + "="*50)

while True:
    question = input("\nYou: ")
    if question.strip().lower() == "exit":
        print("Goodbye! Happy coding!")
        break

    if not question.strip():
        continue

    history_string = "\n".join(history_turns)
    prompt = f"""You are an expert Data Scientist acting as an interactive project guide.
Your goal is to walk the user through building a data science project using the CSV schema provided in the context below.

THE RULES OF CONTINUATION:
1. You must take things ONE Phase at a time. Do not move to the next phase until the user finishes the current task or says they are ready to move on.
2. Structure workflow:
   - Phase 1: Problem Definition & Data Overview (Target variable identification, structural anomalies)
   - Phase 2: Exploratory Data Analysis (EDA) & Data Manipulation (handling missing entries, plotting distributions)
   - Phase 3: Preprocessing & Feature Engineering (scaling numerical columns, encoding categories)
   - Phase 4: Model Selection & Training (traditional ML algorithms or neural networks)
   - Phase 5: Evaluation & Metrics interpretation
3. When giving code snippets, ONLY write Python/Pandas/Scikit-Learn code for the current active phase. Keep it short, practical, and tailored to this dataset schema.
4. Conclude your response by asking the user to run the code, check their console output, and tell you when they are ready for the next step.

Dataset Structural Context:
{csv_info}

Recent Conversation History:
{history_string}

Current Input from User: {question}
Next Mentor Response:"""

    response = llm.invoke(prompt)
    answer = response.content

    print(f"\n🤖 Mentor:\n{answer}")
    print("----------------------")

    new_turn = f"User: {question}\nMentor: {answer}"
    history_turns.append(new_turn)

    if len(history_turns) > MEMORY_LIMIT:
        history_turns.pop(0)
```
